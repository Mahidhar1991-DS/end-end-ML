```python
numeric_feature = [feature for feature in df.columns if df[feature].dtype !='O']
categorical_feature = [feature for feature in df.columns if df[feature].dtype == 'O']

print("we have {} numerical features : {}".format(len(numeric_feature),numeric_feature))
print("we have {} categorical features : {}".format(len(categorical_feature), categorical_feature))
```

```python
df['total_score'] = df['math_score'] + df['reading_score'] + df['writing_score']
df['average_score'] = df['total_score']/3
df.head()
```

```python
reading_full = df[df['reading_score']==100]['average_score'].count()
writing_full = df[df['writing_score'] == 100]['average_score'].count()
math_full = df[df['math_score']==100]['average_score'].count()

print(f'full marks in math: {math_full} ')
print(f'full marks in reading: {reading_full}')
print(f'full marks in writing_full : {writing_full}')
```

```python
reading_full_lessthan_20 = df[df['reading_score'] <=20]['average_score'].count()
writing_full_lessthan_20 = df[df['writing_score'] <=20]['average_score'].count()
math_full_lessthan_20 = df[df['math_score'] <=20]['average_score'].count()


print(f'students scored less than 20 in math: {math_full_lessthan_20}')
print(f'students scored less than 20 in reading: {reading_full_lessthan_20}')
print(f'students scored less than 20 in writing: {writing_full_lessthan_20}')
```

```python
gender_group = df.groupby('gender').mean(numeric_only=True)
gender_group
```

```python
plt.figure(figsize=(10, 8))

X = ['Total Average','Math Average']


female_scores = [gender_group['average_score'][0], gender_group['math_score'][0]]
male_scores = [gender_group['average_score'][1], gender_group['math_score'][1]]

X_axis = np.arange(len(X))
  
plt.bar(X_axis - 0.2, male_scores, 0.4, label = 'Male')
plt.bar(X_axis + 0.2, female_scores, 0.4, label = 'Female')
  
plt.xticks(X_axis, X)
plt.ylabel("Marks")
plt.title("Total average v/s Math average marks of both the genders", fontweight='bold')
plt.legend()
plt.show()
```
